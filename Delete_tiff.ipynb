```python
from pathlib import Path


def delete_tiff_files(folder_path: str, dry_run: bool = True) -> None:


    root = Path(folder_path).expanduser().resolve()

    if not root.exists():
        raise FileNotFoundError(f"Folder does not exist:{root}")

    if not root.is_dir():
        raise NotADirectoryError(f"Path is not a directory: {root}")

    # suffix.lower() 可以同时匹配 .tiff、.TIFF、.Tiff 等形式
    tiff_files = [
        file_path
        for file_path in root.rglob("*")
        if file_path.is_file() and file_path.suffix.lower() == ".tiff"
    ]

    if not tiff_files:
        print("No .tiff files found.")
        return

    print(f"Found {len(tiff_files)} .tiff files:")

    deleted_count = 0
    failed_count = 0

    for file_path in tiff_files:
        if dry_run:
            print(f"[Preview] {file_path}")
            continue

        try:
            file_path.unlink()
            deleted_count += 1
            print(f"[Deleted] {file_path}")
        except OSError as error:
            failed_count += 1
            print(f"[Failed to delete] {file_path}\nReason: {error}")

    if dry_run:
        print("\nCurrent mode is preview, no files actually deleted.")
        print("After confirmation, change dry_run to False.")
    else:
        print(
            f"\nProcessing complete: Successfully deleted {deleted_count} files, "
            f"failed to delete {failed_count} files."
        )


```
